```python
import numpy as np

docs = [
    'go india',
    'india india',
    'hip hip hurray',
    'jeetega bhai jeetega india jeetega',
    'bharat mata ki jai',
    'kohli kohli',
    'sachin sachin',
    'dhoni dhoni',
    'modi ji ki jai',
    'inquilab zindabad',
    'artificial intelligence and machine learning are transforming technology',
    'the quick brown fox jumps over the lazy dog',
    'sunny days at the beach are the absolute best',
    'exploring new cities and trying local street food is amazing',
    'regular exercise and a healthy diet keep the mind sharp',
    'spacex is launching another rocket into orbit tonight',
    'reading books expands your imagination and knowledge'
]
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
# Initialize the Tokenizer
tokenizer = Tokenizer(oov_token = '<Chunchunmaru>')

# Fit the tokenizer on our document list
tokenizer.fit_on_texts(docs)

# Display results
print(f"Total unique words (vocabulary size): {len(tokenizer.word_index)}")
print("\nWord Index:")
print(tokenizer.word_index)

print("\nWord Counts:")
print(dict(list(tokenizer.word_counts.items())[:10]))  # Display first 10 items
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Convert documents to sequences of integers
sequences = tokenizer.texts_to_sequences(docs)

# Pad sequences so they all have the same length
padded_sequences = pad_sequences(sequences, padding='post')

print("Original Document sample:")
print(docs[0])

print("\nInteger Sequence:")
print(sequences[0])

print("\nPadded Sequence:")
print(padded_sequences[0])

print("\nShape of padded data:")
print(padded_sequences.shape)
```

```python
from tensorflow.keras.datasets import imdb
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense ,SimpleRNN ,Flatten,Embedding
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train = pad_sequences(X_train,padding = 'post',maxlen = 100)
X_test = pad_sequences(X_test,padding = 'post',maxlen = 100)
```

```python
X_train[0]
```

```python
model = Sequential()

model.add(SimpleRNN(64,input_shape = (100,1),return_sequences = False))
model.add(Dense(1,activation = 'sigmoid'))

model.summary()
```

```python
model.compile(loss = 'binary_crossentropy',optimizer = 'adam',metrics = ['accuracy'])

model.fit(X_train,y_train,epochs = 10,validation_data = (X_test,y_test))
#
```
